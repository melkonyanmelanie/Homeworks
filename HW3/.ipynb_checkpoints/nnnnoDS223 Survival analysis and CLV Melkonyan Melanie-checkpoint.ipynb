```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from lifelines import (
    WeibullAFTFitter, LogLogisticAFTFitter, LogNormalAFTFitter,
    GeneralizedGammaRegressionFitter, PiecewiseExponentialRegressionFitter
)
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('telco.csv')
df['churn'] = df['churn'].map({'Yes': 1, 'No': 0})
df['gender'] = df['gender'].map({'Female': 0, 'Male': 1})
df['marital'] = df['marital'].map({'Married': 1, 'Unmarried': 0})
for col in ['retire', 'voice', 'internet', 'forward']:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

# One-hot encode categorical variables 
categorical_columns = ['region', 'ed', 'custcat']
df_encoded = pd.get_dummies(df, columns=categorical_columns, drop_first=True)

features = [col for col in df_encoded.columns if col not in ['ID', 'tenure', 'churn']]

duration_col = 'tenure'
event_col = 'churn'
```

```python
weibull_aft = WeibullAFTFitter()
weibull_aft.fit(df_encoded[features + [duration_col, event_col]], duration_col=duration_col, event_col=event_col)
print("Weibull AFT model summary:")
print(weibull_aft.summary)
```

```python
loglogistic_aft = LogLogisticAFTFitter()
loglogistic_aft.fit(df_encoded[features + [duration_col, event_col]], duration_col=duration_col, event_col=event_col)
print("Log-Logistic AFT model summary:")
print(loglogistic_aft.summary)
```

```python
lognormal_aft = LogNormalAFTFitter()
lognormal_aft.fit(df_encoded[features + [duration_col, event_col]], duration_col=duration_col, event_col=event_col)
print("Log-Normal AFT model summary:")
print(lognormal_aft.summary)
```

```python
# Scale tenure
df_encoded['tenure'] = df_encoded['tenure'] / 100

# Add penalizer and use alternate optimizer
gen_gamma_reg = GeneralizedGammaRegressionFitter(penalizer=0.01)
gen_gamma_reg._scipy_fit_method = "SLSQP"
gen_gamma_reg.fit(df_encoded[features + ['tenure', event_col]], duration_col='tenure', event_col=event_col)
print(gen_gamma_reg.summary)
```

```python
pe_exp_reg = PiecewiseExponentialRegressionFitter(breakpoints=[12, 24, 36, 48])
pe_exp_reg.fit(df_encoded[features + [duration_col, event_col]], duration_col=duration_col, event_col=event_col)
print("Piecewise Exponential Regression model summary:")
print(pe_exp_reg.summary)
```
