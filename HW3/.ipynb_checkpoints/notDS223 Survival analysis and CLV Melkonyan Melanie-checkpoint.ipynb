```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from lifelines import (
    AalenJohansenFitter, BreslowFlemingHarringtonFitter, ExponentialFitter, GeneralizedGammaFitter,
    KaplanMeierFitter, LogLogisticFitter, LogNormalFitter, MixtureCureFitter, NelsonAalenFitter,
    PiecewiseExponentialFitter, SplineFitter, WeibullFitter
)
```

```python
df = pd.read_csv('telco.csv')
df.columns = [c.strip().lower() for c in df.columns]
df['churn'] = df['churn'].map({'Yes': 1, 'No': 0})

# List all yes/no columns 
yes_no_cols = ['retire', 'voice', 'internet', 'forward']  
for col in yes_no_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df['gender'] = df['gender'].map({'Female': 0, 'Male': 1})

print(df.head())
print("Missing values by column:", df.isnull().sum().to_dict())
```

```python
T = df['tenure']
E = df['churn']

plt.figure(figsize=(16,10))
models = []

models.append(('AalenJohansen', AalenJohansenFitter()))
models.append(('BreslowFlemingHarrington', BreslowFlemingHarringtonFitter()))
models.append(('Exponential', ExponentialFitter()))
models.append(('GeneralizedGamma', GeneralizedGammaFitter()))
models.append(('KaplanMeier', KaplanMeierFitter()))
models.append(('LogLogistic', LogLogisticFitter()))
models.append(('LogNormal', LogNormalFitter()))
models.append(('MixtureCure', MixtureCureFitter(base_fitter=WeibullFitter())))
models.append(('NelsonAalen', NelsonAalenFitter()))
models.append(('PiecewiseExponential', PiecewiseExponentialFitter([1,6,12,24,36,48])))  # tune breaks
models.append(('Spline', SplineFitter(knot_locations=[12,24,36,48])))
models.append(('Weibull', WeibullFitter()))
```

```python
for name, model in models:
    try:
        if name == 'AalenJohansen':
            model.fit(T, E, event_of_interest=1)
            model.plot(label=name)
        elif name == 'MixtureCure':
            model.fit(T, E)
            model.plot_survival_function(label=name)
        elif name == 'NelsonAalen':
            model.fit(T, E)
            model.plot_cumulative_hazard(label=name)
        else:
            model.fit(T, E)
            model.plot_survival_function(label=name)
    except Exception as ex:
        print(f"{name} failed: {ex}")

plt.title("Survival Curve Comparison: Telco Churn")
plt.xlabel("Tenure (months)")
plt.ylabel("Survival Probability")
plt.legend()
plt.show()
```
